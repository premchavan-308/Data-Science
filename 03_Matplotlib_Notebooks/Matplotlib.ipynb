```python
import matplotlib.pyplot as plt
```

## Create First Chart

```python
x=[1,2,3,4,5]
y=[10,20,15,35,30]

plt.plot(x,y)
plt.show()
```

## Customize Chart

```python
plt.figure(figsize=(4,3))
plt.plot(x,y,color='red',marker='o',linestyle='--',linewidth=2,markersize=8)

plt.title("Daily Sales")
plt.xlabel("Months")
plt.ylabel("Cost")
plt.show()
```

## Advanced- Multiple lines and legends

```python
x = [1,2,3,4,5]
y1 = [10, 40, 25, 35, 15]
y2 = [20, 10, 35, 25, 45]

plt.plot(x,y1,label='Sales 2024')
plt.plot(x,y2,label='Sales 2025')

plt.title('Yearly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.legend()
plt.show()
```

## Bar Chart

```python
x = ['A','B','C','D','E']
y = [10, 20, 55, 35, 45]

plt.bar(x,y)
plt.title('Bar Chart Example')
plt.show()
```

## Histogram

```python
import random
data=[random.randint(1,20) for _ in range(300)]

plt.hist(data,bins=20)
plt.title('Histogram Example')
plt.show()
```

## Pie Chart

```python
categories = ['A','B','C','D','E']
sales = [10, 20, 55, 35, 45]

plt.pie(sales, labels=categories, autopct = '%1.1f%%', startangle=90)
plt.title("Pie Chart Example")
plt.show()
```

## Scatter plot

```python
y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]

plt.scatter(y1, y2)
plt.title("Scatter Plot Example")
plt.show()
```

## Subplots

```python
categories=['Mon','Tue','Wed','Thr','Fri']
sales=[10,20,55,35,45]

y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.bar(categories,sales)
plt.title('Daily Sales')
plt.xlabel('Weekly')
plt.ylabel('Sales')

plt.subplot(1,2,2)
plt.scatter(y1,y2)
plt.title('User Example')
plt.xlabel('User1')
plt.ylabel('User2')

plt.show()
```

## Matplotlib with Pandas

```python
import pandas as pd
```

```python
data={
    'Months' : ['JAN','FEB','MAR','APR','MAY','JUN'],
    'Sales' : [11000,44000,20000,70000,34000,59000]
}
df=pd.DataFrame(data)
df
```

```python
plt.bar(df['Months'],df['Sales'])
plt.title('Monthly_Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

## Save the image of chart

```python
plt.bar(df['Months'], df['Sales'])
plt.title("Matplotlib with Pandas")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.savefig("Monthly_Sales_from_df.png")

plt.show()
```
