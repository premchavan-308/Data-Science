```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Matplotlib Basic

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("Basic Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

## Subplots

```python
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,1)
plt.plot(x**2,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x)
```

## Object oriented

```python
fig=plt.figure(figsize=(20,8),dpi=100)
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2=fig.add_axes([0.1,0.3,0.4,0.4])
axis2.plot(x,y)
```

## Types of plot

```python
plt.scatter(x,y)
```
