```python
df1 = pd.DataFrame({'ID':[1,2,3],'Name':['A','B','C']})          #1ST DATAFRAME
df1
```

```python
df2 = pd.DataFrame({'ID':[1,2,2,4],'Score':[88,96,77,79]})          #2ND DATAFRAME
df2
```

```python
pd.concat([df1,df2], axis=0)          #MERGE OR JOIN ONE-AFTER-OTHER BY ROW
```

```python
pd.concat([df1,df2], axis=1)           #MERGE OR JOIN SIDE-BY-SIDE BY COLUMN
```

```python
pd.merge(df1,df2, how='inner', on='ID')         #USE INNER JOIN
```

```python
pd.merge(df1,df2, how='inner', left_on='ID', right_on='ID')          #USE LEFT AND RIGHT INNER JOIN
```

```python
pd.merge(df1,df2, how='outer', on='ID')         #USE OUTER JOIN ON DATAFRAME
```
